```python
from jupyturtle import forward, left, make_turtle, penup, pendown, right
def parallelogram(wide, tall, angle):
    forward(wide)
    left(angle)
    forward(tall)
    left(180-angle)
    forward(wide)
    left(angle)
    forward(tall)

make_turtle()
parallelogram(100, 50, 90)
penup()
right(90)
forward(20)
pendown()
parallelogram(50, 50, 60)
penup()
left(30)
forward(50)
left(90)
pendown()
parallelogram(100, 50, 60)
```
