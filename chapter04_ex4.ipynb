```python
from jupyturtle import left, right, forward, make_turtle
import math
def triangle(length, n):
    angle1 = 180 - 360 / (2* n)
    angle2 = 360 / n
    forward(length)
    left(angle1)
    forward((length * -math.cos(360 / (2 * n))) * 2)
    left(angle1)
    forward(length)
    left(angle2)
make_turtle()
triangle(50, 3)
```
